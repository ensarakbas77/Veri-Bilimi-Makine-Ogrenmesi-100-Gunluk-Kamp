### **Understanding Data Encoding in Feature Engineering (Beginner-Friendly)**  
### **Özellik Mühendisliğinde Veri Kodlamayı Anlama (Başlangıç Seviyesi)**

When working with machine learning models, we often encounter categorical data—data that consists of labels or categories instead of numerical values. However, machine learning models work best with numbers, so we need to **convert categorical data into numerical form** using **data encoding** techniques.  
Makine öğrenmesi modelleri ile çalışırken sık sık kategorik verilerle karşılaşırız—bu veriler sayılar yerine etiketler veya kategorilerden oluşur. Ancak makine öğrenmesi modelleri sayılarla en iyi şekilde çalışır, bu yüzden **kategorik verileri sayısal forma dönüştürmemiz** gerekir. Bunu **veri kodlama** teknikleri ile yaparız.

## **Why is Data Encoding Important?**  
## **Veri Kodlama Neden Önemlidir?**
- Machine learning models require numerical input, and categorical values must be transformed into numbers.  
  Makine öğrenmesi modelleri sayısal girdi gerektirir, bu yüzden kategorik değerler sayılara dönüştürülmelidir.
- Proper encoding helps models understand relationships between categories.  
  Doğru kodlama, modellerin kategoriler arasındaki ilişkileri anlamasına yardımcı olur.
- Different encoding techniques are suitable for different types of categorical data.  
  Farklı kodlama teknikleri, farklı kategorik veri türleri için uygundur.

## **Types of Data Encoding**  
## **Veri Kodlama Türleri**
There are multiple encoding techniques, each suited to different situations.  
Farklı durumlara uygun birden fazla kodlama tekniği vardır.

### **1. One-Hot Encoding**  
### **1. Tek-Sıcak Kodlama (One-Hot Encoding)**
- Converts each unique category into a separate binary (0 or 1) column.  
  Her benzersiz kategoriyi ayrı bir ikili (0 veya 1) sütuna dönüştürür.
- Best for **nominal (unordered) categories** with a small number of unique values.  
  Az sayıda benzersiz değere sahip **nominal (sırasız) kategoriler** için uygundur.

#### **Example / Örnek:**
| Color  | One-Hot Encoding |
|--------|-----------------|
| Red    | (1,0,0)         |
| Blue   | (0,1,0)         |
| Green  | (0,0,1)         |

```python
import pandas as pd
df = pd.DataFrame({'Color': ['Red', 'Blue', 'Green']})
df_encoded = pd.get_dummies(df, columns=['Color'])
print(df_encoded)
```

📌 **Use When:** The categorical variable has a small number of unique values.  
📌 **Ne Zaman Kullanılır:** Kategorik değişken az sayıda benzersiz değere sahipse.

---

### **2. Label Encoding**  
### **2. Etiket Kodlama (Label Encoding)**
- Assigns each category a unique integer.  
  Her kategoriye benzersiz bir tamsayı atar.
- Suitable for **ordinal categories** where order matters (e.g., Low, Medium, High).  
  Sıralamanın önemli olduğu **ordinal kategoriler** için uygundur (ör. Düşük, Orta, Yüksek).

#### **Example / Örnek:**
| Size   | Label Encoding |
|--------|---------------|
| Small  | 0             |
| Medium | 1             |
| Large  | 2             |

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
df['Size_Encoded'] = encoder.fit_transform(df['Size'])
```

📌 **Use When:** There is a natural order between categories.  
📌 **Ne Zaman Kullanılır:** Kategoriler arasında doğal bir sıralama varsa.

---

### **3. Ordinal Encoding**  
### **3. Sıralı Kodlama (Ordinal Encoding)**
- Similar to Label Encoding but allows us to **manually specify the order**.  
  Label Encoding’e benzer ancak **sıralamayı manuel olarak belirlememizi** sağlar.

#### **Example / Örnek (Manually Assigning Order / Sıralamayı Manuel Belirleme):**
| Education | Ordinal Encoding |
|-----------|-----------------|
| High School | 1             |
| Bachelor's  | 2             |
| Master's    | 3             |

```python
from sklearn.preprocessing import OrdinalEncoder
encoder = OrdinalEncoder(categories=[['High School', "Bachelor's", "Master's"]])
df['Education_Encoded'] = encoder.fit_transform(df[['Education']])
```

📌 **Use When:** The categories follow a meaningful ranking.  
📌 **Ne Zaman Kullanılır:** Kategoriler anlamlı bir sıralamaya sahipse.

---

### **4. Frequency Encoding**  
### **4. Frekans Kodlama (Frequency Encoding)**
- Replaces each category with its frequency (count) in the dataset.  
  Her kategoriyi veri kümesindeki frekansı (sayısı) ile değiştirir.
- Helps when there are many unique categories.  
  Çok sayıda benzersiz kategori olduğunda yardımcı olur.

#### **Example / Örnek:**
| City     | Count | Frequency Encoding |
|----------|-------|-------------------|
| New York | 100   | 100               |
| London   | 80    | 80                |
| Tokyo    | 120   | 120               |

```python
df['City_Freq'] = df['City'].map(df['City'].value_counts())
```

📌 **Use When:** The category's frequency might be an important feature.  
📌 **Ne Zaman Kullanılır:** Kategorinin frekansı önemli bir özellik olabilecekse.

---

### **5. Target Encoding (Mean Encoding)**  
### **5. Hedef Kodlama (Ortalama Kodlama)**
- Replaces categories with the mean of the target variable.  
  Kategorileri hedef değişkenin ortalaması ile değiştirir.
- Best for situations where the categorical feature has a strong correlation with the target.  
  Kategorik özelliğin hedef ile güçlü bir korelasyonu olduğu durumlarda uygundur.

#### **Example / Örnek:**
| Product | Target (Sales) | Mean Encoding |
|---------|--------------|--------------|
| A       | 100          | 120          |
| A       | 140          | 120          |
| B       | 200          | 200          |

```python
df['Product_Encoded'] = df.groupby('Product')['Sales'].transform('mean')
```

📌 **Use When:** The categorical feature influences the target variable.  
📌 **Ne Zaman Kullanılır:** Kategorik özellik hedef değişkeni etkiliyorsa.

---

### **6. Binary Encoding**  
### **6. İkili Kodlama (Binary Encoding)**
- Converts categories into binary format and encodes them into separate columns.  
  Kategorileri ikili (binary) formata çevirip ayrı sütunlara kodlar.
- More memory-efficient than One-Hot Encoding for high-cardinality categorical data.  
  Yüksek kardinaliteli (çok benzersiz değerli) kategorik verilerde One-Hot Encoding’den daha verimlidir.

#### **Example / Örnek (for 4 categories / 4 kategori için):**
| Category | Binary | Encoded Columns |
|----------|--------|----------------|
| A        | 00     | (0,0)          |
| B        | 01     | (0,1)          |
| C        | 10     | (1,0)          |
| D        | 11     | (1,1)          |

```python
from category_encoders import BinaryEncoder
encoder = BinaryEncoder(cols=['Category'])
df_encoded = encoder.fit_transform(df)
```

📌 **Use When:** The dataset has many unique categories (high cardinality).  
📌 **Ne Zaman Kullanılır:** Veri kümesinde çok sayıda benzersiz kategori varsa.

---

### **Choosing the Right Encoding Technique**  
### **Doğru Kodlama Tekniğini Seçmek**
| Encoding Type       | Suitable For              | Pros                          | Cons                          |
|---------------------|--------------------------|-------------------------------|-------------------------------|
| **One-Hot Encoding** | Small categorical values (Nominal) | Simple, widely used | Creates too many columns if categories are large |
| **Label Encoding**  | Ordered categories (Ordinal) | Efficient, small memory | Can mislead models into thinking numbers imply order |
| **Ordinal Encoding** | Categories with order | Respects order | Requires prior knowledge of order |
| **Frequency Encoding** | Many unique categories | Compact representation | Can lead to overfitting |
| **Target Encoding** | Categorical with strong target correlation | Keeps relationship with target | Can introduce data leakage |
| **Binary Encoding** | High-cardinality categories | Efficient with many categories | Not as interpretable |

| Kodlama Türü        | Uygun Olduğu Durumlar     | Avantajları                   | Dezavantajları                |
|---------------------|--------------------------|-------------------------------|-------------------------------|
| **One-Hot Encoding** | Küçük kategorik değerler (Nominal) | Basit, yaygın | Kategori sayısı büyükse çok fazla sütun oluşturur |
| **Label Encoding**  | Sıralı kategoriler (Ordinal) | Verimli, az bellek kullanır | Sayılar sıralama anlamı taşıyormuş gibi modele yanlış bilgi verebilir |
| **Ordinal Encoding** | Sıralı kategoriler | Sıralamayı korur | Önceden sıralama bilgisinin bilinmesi gerekir |
| **Frequency Encoding** | Çok sayıda benzersiz kategori | Kompakt gösterim | Aşırı öğrenmeye (overfitting) neden olabilir |
| **Target Encoding** | Hedefle güçlü korelasyonlu kategoriler | Hedefle ilişkiyi korur | Veri sızıntısına yol açabilir |
| **Binary Encoding** | Yüksek kardinaliteli kategoriler | Verimli | Yorumlamak zordur |

### **Final Tip / Son Tavsiye:**
Before choosing an encoding method, **analyze the data** and **understand the nature of the categories**. The right encoding technique can **improve model performance** while avoiding unnecessary complexity.  
Bir kodlama yöntemi seçmeden önce **veriyi analiz edin** ve **kategorilerin doğasını anlayın**. Doğru kodlama tekniği, gereksiz karmaşıklıktan kaçınarak **model performansını artırabilir**.
