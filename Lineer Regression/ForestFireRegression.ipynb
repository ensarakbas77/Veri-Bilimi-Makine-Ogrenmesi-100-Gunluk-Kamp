```python
# dependent & independent features

X = df.drop("FWI", axis=1)
y = df["FWI"]

X.head()
```

```python
# train - test split
from sklearn.model_selection import train_test_split
 
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=15)
```

```python
print("X Train:",X_train.shape)
print("X Test:",X_test.shape)
print("y Train:",y_train.shape)
print("y Test:",y_test.shape)
```

```python
# İki tane feature ın korelasyonu çok yüksekse ikisinden biri çıkartılabilir.
# redundancy = iki farklı kolonun birbiriyle çok bağımlı olması, modelin öğrenme sürecine katkısı yok, karmaşıklığı artırabilir.
# multicollinearity 

X_train.corr()
```

```python
print(X_train.corr().iloc[0,2])
```

```python
def correlation_for_dropping(df, threshold):
    columns_to_drop = set()
    corr = df.corr()
    for i in range(len(corr.columns)):
        for j in range(i):
           if abs(corr.iloc[i, j]) > threshold:
               columns_to_drop.add(corr.columns[i])

    return columns_to_drop
```

```python
columns_dropping = correlation_for_dropping(X_train, 0.85)
columns_dropping
```

```python
X_train.drop(columns_dropping, axis=1, inplace=True)
X_test.drop(columns_dropping, axis=1, inplace=True)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
# scaler

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
plt.subplots(figsize=(15,5))
plt.subplot(1,2,1)
sns.boxplot(data=X_train)
plt.title("X Train")
plt.subplot(1,2,2)
sns.boxplot(data=X_train_scaled)
plt.title("X Train Scaled")
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
linear = LinearRegression()
linear.fit(X_train_scaled, y_train)
y_pred = linear.predict(X_test_scaled)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
score = r2_score(y_test, y_pred)

print("Mean Absolute Error: ", mae)
print("Mean Squared Error: ", mse)
print("R2 Squared: ", score)

plt.scatter(y_test, y_pred)
plt.show()
```

```python
from sklearn.linear_model import Lasso

lasso = Lasso()
lasso.fit(X_train_scaled, y_train)
y_pred = lasso.predict(X_test_scaled)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
score = r2_score(y_test, y_pred)

print("Mean Absolute Error: ", mae)
print("Mean Squared Error: ", mse)
print("R2 Squared: ", score)

plt.scatter(y_test, y_pred)
plt.show()
```
